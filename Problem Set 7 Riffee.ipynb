At α = .05, the interval is [-1.960, 1.960], so our z score for X_bar (=2) is outside of this window. We reject the null at the 95% confidence level.  

At α = .01, the interval is [-2.576, 2.576], , so our z score for X_bar (=2) is within this this window. We do NOT reject the null at the 99% confidence level.
