```python
p7 = pd.read_csv('./data/ames_prices.csv')
```

```python
p7['Central.Air'].unique()
```

```python
air = pd.get_dummies(p7['Central.Air'], drop_first = True, dtype = int)
air
```

**7. A**

```python
p_hat = sum(air.loc[:,'Y'])/len(air.loc[:,'Y'])
p_hat
```

**7 B.**

```python
#The number of bootstrap repetitions, R, will be 5,000
R = 5_000

#The seed I used for this random number sequence is 100 
rng = np.random.default_rng(100)

#each p_hat is the bootstrap probability of a home having central air. 
p_hats = []

for x in range(R):
    boot_sample = rng.choice(air, size = len(air.loc[:,'Y']), replace = True)
    boot_p_hat = sum(boot_sample)/len(boot_sample)

    p_hats.append(boot_p_hat)

```

**7. C.**

```python
n = len(p_hats)
p_hat_mean = sum(p_hats)/len(p_hats)
sum_squared_diffs = sum((x - p_hat_mean)**2 for x in p_hats)
```

```python
SE_p_hats = np.sqrt(1/(n-1) * sum_squared_diffs)
SE_p_hats
```

```python
#Change the output from an 1-value array to a float 
SE_p_hats = SE_p_hats[0]
SE_p_hats
```

**7. D**

```python
#p_hat is the original estimate
#SE_p_hats is the bootstrapped standard error

#90% CI:
#+/-1.645 * SE + p_hat = critical values
crit_value1 = -1.645 * SE_p_hats + p_hat
crit_value2 = 1.645 * SE_p_hats + p_hat

print(f'The 90% confidence interval for the porportion of homes that have central heating is {crit_value1.round(4)}, {crit_value2.round(4)}')
```

```python
#95% CI:
#+/-1.960 * SE + p_hat = critical values
crit_value1 = -1.960 * SE_p_hats + p_hat
crit_value2 = 1.960 * SE_p_hats + p_hat

print(f'The 95% confidence interval for the porportion of homes that have central heating is {crit_value1.round(4)}, {crit_value2.round(4)}')
```

```python
#99% CI:
#+/-2.576 * SE + p_hat = critical values
crit_value1 = -2.576 * SE_p_hats + p_hat
crit_value2 = 2.576 * SE_p_hats + p_hat

print(f'The 99% confidence interval for the porportion of homes that have central heating is {crit_value1.round(4)}, {crit_value2.round(4)}')
```

**7. E**

Given the null hypothesis that 90% of homes have central heating (i.e., H_0: p = 0.9), we:
- reject the null at the 90% confidence level (i.e., alpha = 0.1), because our confidence interval is: [0.9255, 0.9407]
- reject the null at the 95% confidence level (i.e., alpha = 0.05), because our confidence interval is: [0.9241, 0.9421]
- reject the null at the 99% confidence level (i.e., alpha = 0.01), because our confidence interval is: [0.9212, 0.945]

These results strongly suggest that the true population porportion of homes that have central air is greater than 90%, because even our widest confidence interval at alpha = 0.01 ranges from 0.9212, 0.945.

```python
area = p7.loc[:,'area']
price = p7.loc[:,'price']
```

```python
pearson = scipy.stats.pearsonr(np.arcsinh(area), np.arcsinh(price))
pearson[0] #I can get just the pearson correlation out by indexing the output for the first number.
```

**8. A**

```python
#The number of bootstrap repetitions, R, will be 2,000
R = 2_000

#The seed I used for this random number sequence is 100 
rng = np.random.default_rng(100)
n = len(p7.loc[:,'area'])

#To be filled in with pearson coefficients. 
pearsons = []

for x in range(R):
    boot_index = rng.choice(n, size = n, replace = True)
    boot_area = area[boot_index]
    boot_price = price[boot_index]
    boot_pearson = scipy.stats.pearsonr(np.arcsinh(boot_area), np.arcsinh(boot_price))
    pearsons.append(boot_pearson[0])
    
```

**8. B**

```python
def boot_SE_calculator(bootstrap_values):
    n = len(bootstrap_values)
    mean_boot_value = sum(bootstrap_values) / n
    sum_squared_diffs = sum((x -mean_boot_value)**2 for x in bootstrap_values)
    SE_bootstrap = np.sqrt(1/(n-1) * sum_squared_diffs)
    return SE_bootstrap
    
```

```python
pearsons_SE = boot_SE_calculator(pearsons)
```

```python
print(f'the bootstrap standard error for the pearson correlation between np.arcsinh(area)and np.arcsinh(price) of houses is {boot_SE_calculator(pearsons).round(4)}')
```

**8. C**

```python
def normal_crit_values_calculator(estimate, standard_error):
    #90% CI:
    #+/-1.645 * SE + p_hat = critical values
    crit_value1_90 = -1.645 * standard_error + estimate
    crit_value2_90 = 1.645 * standard_error + estimate

    #95% CI:
    #+/-1.960 * SE + p_hat = critical values
    crit_value1_95 = -1.960 * standard_error + estimate
    crit_value2_95 = 1.960 * standard_error + estimate

    #99% CI:
    #+/-2.576 * SE + p_hat = critical values
    crit_value1_99 = -2.576 * standard_error + estimate
    crit_value2_99 = 2.576 * standard_error + estimate

    print(f'The 90% confidence interval is {crit_value1_90.round(4)}, {crit_value2_90.round(4)}')
    print(f'The 95% confidence interval is {crit_value1_95.round(4)}, {crit_value2_95.round(4)}')
    print(f'The 99% confidence interval is {crit_value1_99.round(4)}, {crit_value2_99.round(4)}')
    
```

```python
normal_crit_values_calculator(pearson[0], pearsons_SE)
```

Given the null hypothesis that the area of a home is NOT correlated with the price of a home (i.e., H_0: the peason correlation = 0), we reject the null hypothesis at the alpha = 0.1, 0.05, and 0.01 values, because the confidence interval for each of these alphas does not contain 0. 

In other words, this strongly suggests that the price and the area of a home are correlated, and that they are relatively strongly correlated. For instance, our 99% confidence interval for pearson correlation is [0.6971, 0.7493]. While this does not mean that the price of the home is determined by area, it does mean that homes with higher area tend to also have higher prices. 

**8. E.**

For any given row, the area and the price both refer to the area and price of the same house. We need to keep area and price paired together during resampling so that we keep the price of a house associated with the area of that house. If we do not keep that association, our data will become meaningless. 

I did this by bootstrapping the row index, and then using the bootstrapped row index to pull corresponding area and price observations for each bootstrapped house. Then, I computed the pearson correlation.

```python
def my_bootstrapper(data):
    '''Gonna make a bootstrap function for myself. 'data' should be in list form for this function. Will return a list of the mean result for each bootstrap iteration.'''
    n = len(data)
    R = 2_000
    rng = np.random.default_rng(100)
    boot_outputs = []

    for x in range(R):
        boot_result = rng.choice(data, size = n, replace = True)
        boot_mean = sum(boot_result) / n
        boot_outputs.append(boot_mean)

    return boot_outputs
```
